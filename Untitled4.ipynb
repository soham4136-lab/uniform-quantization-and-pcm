```python
import numpy as np
import matplotlib.pyplot as plt

# ============================================================
# 1. Generate a full-scale sinusoidal signal
# ============================================================

fs = 10000                 # Sampling frequency
f = 100                    # Signal frequency
duration = 1               # seconds

t = np.arange(0, duration, 1/fs)

# Full-scale sinusoid: amplitude = 1
x = np.sin(2 * np.pi * f * t)


# ============================================================
# 2. Uniform PCM Quantizer from first principles
# ============================================================

def uniform_pcm_quantizer(x, bits, xmin=-1, xmax=1):
    """
    Uniform PCM quantizer.

    Returns:
        xq       : quantized samples
        indices  : quantization indices
        codewords: binary PCM codewords
        mse      : mean squared error
        sqnr_db  : SQNR in dB
    """

    L = 2 ** bits

    # Quantization step size
    delta = (xmax - xmin) / L

    # --------------------------------------------------------
    # Clip input to allowed quantizer range
    # --------------------------------------------------------
    x_clip = np.clip(x, xmin, xmax)

    # --------------------------------------------------------
    # Calculate quantization index
    # --------------------------------------------------------
    indices = np.floor((x_clip - xmin) / delta).astype(int)

    # Handle the xmax boundary
    indices = np.minimum(indices, L - 1)

    # --------------------------------------------------------
    # Calculate quantized value
    # Using midpoint reconstruction levels
    # --------------------------------------------------------
    xq = xmin + (indices + 0.5) * delta

    # --------------------------------------------------------
    # Convert indices to binary codewords
    # --------------------------------------------------------
    codewords = [
        format(index, f'0{bits}b')
        for index in indices
    ]

    # --------------------------------------------------------
    # MSE
    # --------------------------------------------------------
    error = x - xq
    mse = np.mean(error ** 2)

    # --------------------------------------------------------
    # SQNR
    # --------------------------------------------------------
    signal_power = np.mean(x ** 2)
    noise_power = mse

    sqnr_db = 10 * np.log10(signal_power / noise_power)

    return xq, indices, codewords, mse, sqnr_db


# ============================================================
# 3. Quantize using 2 to 8 bits
# ============================================================

results = []

for bits in range(2, 9):

    xq, indices, codewords, mse, sqnr_db = \
        uniform_pcm_quantizer(x, bits)

    # Standard theoretical approximation
    theoretical_sqnr = 6.02 * bits + 1.76

    results.append([
        bits,
        2 ** bits,
        mse,
        sqnr_db,
        theoretical_sqnr
    ])


# ============================================================
# 4. Display results
# ============================================================

print("=" * 78)
print("Uniform PCM Quantizer Performance")
print("=" * 78)

print(f"{'Bits':>6} {'Levels':>8} {'MSE':>15} "
      f"{'Measured SQNR':>18} {'Theory SQNR':>18}")

print("-" * 78)

for row in results:
    print(f"{row[0]:>6} {row[1]:>8} {row[2]:>15.8e} "
          f"{row[3]:>18.4f} {row[4]:>18.4f}")


# ============================================================
# 5. Show quantized samples, indices and binary codewords
# ============================================================

bits_demo = 3

xq, indices, codewords, mse, sqnr_db = \
    uniform_pcm_quantizer(x, bits_demo)

print("\n")
print("=" * 70)
print(f"Example PCM Output for {bits_demo}-bit Quantizer")
print("=" * 70)

print(f"{'Sample':>8} {'Input':>12} {'Index':>10} "
      f"{'Quantized':>15} {'Codeword':>12}")

print("-" * 70)

# Display first 20 samples
for i in range(20):
    print(f"{i:>8} {x[i]:>12.5f} {indices[i]:>10} "
          f"{xq[i]:>15.5f} {codewords[i]:>12}")


# ============================================================
# 6. Plot original and quantized signal
# ============================================================

plt.figure(figsize=(12, 5))

samples_to_plot = 200

plt.plot(
    t[:samples_to_plot],
    x[:samples_to_plot],
    label="Original Signal"
)

plt.step(
    t[:samples_to_plot],
    xq[:samples_to_plot],
    where='mid',
    label=f"{bits_demo}-bit Quantized Signal"
)

plt.xlabel("Time (s)")
plt.ylabel("Amplitude")
plt.title("Uniform PCM Quantization")
plt.grid(True)
plt.legend()
plt.show()


# ============================================================
# 7. Plot MSE vs number of bits
# ============================================================

bits_array = np.array([r[0] for r in results])
mse_array = np.array([r[2] for r in results])

plt.figure(figsize=(8, 5))

plt.semilogy(
    bits_array,
    mse_array,
    'o-'
)

plt.xlabel("Number of Bits")
plt.ylabel("MSE")
plt.title("MSE vs Number of Quantization Bits")
plt.grid(True)
plt.xticks(bits_array)

plt.show()


# ============================================================
# 8. Compare measured SQNR with theoretical approximation
# ============================================================

measured_sqnr = np.array([r[3] for r in results])
theoretical_sqnr = np.array([r[4] for r in results])

plt.figure(figsize=(8, 5))

plt.plot(
    bits_array,
    measured_sqnr,
    'o-',
    label="Measured SQNR"
)

plt.plot(
    bits_array,
    theoretical_sqnr,
    's--',
    label="Theoretical: 6.02N + 1.76 dB"
)

plt.xlabel("Number of Bits")
plt.ylabel("SQNR (dB)")
plt.title("Measured vs Theoretical SQNR")
plt.grid(True)
plt.xticks(bits_array)
plt.legend()

plt.show()


# ============================================================
# 9. Print SQNR comparison
# ============================================================

print("\n")
print("=" * 65)
print("SQNR Comparison")
print("=" * 65)

print(f"{'Bits':>6} {'Measured SQNR':>20} "
      f"{'Theory (6.02N+1.76)':>25} {'Difference':>15}")

print("-" * 65)

for i, bits in enumerate(bits_array):

    difference = measured_sqnr[i] - theoretical_sqnr[i]

    print(
        f"{bits:>6} "
        f"{measured_sqnr[i]:>20.4f} "
        f"{theoretical_sqnr[i]:>25.4f} "
        f"{difference:>15.4f}"
    )
```
